```python
import operator
from typing import TypedDict, List, Annotated 
from pydantic import BaseModel, Field

from langgraph.graph import StateGraph, START, END
from langgraph.types import Send
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.messages import SystemMessage, HumanMessage

load_dotenv()
```

```python
class Task(BaseModel):
    id: int
    title: str
    brief: str = Field(..., description= "What to cover")
```

```python
class Plan(BaseModel):
    blog_title: str
    tasks: List[Task]
```

```python
class State(TypedDict):
    topic: str
    plan : Plan
    sections : Annotated[list[str], operator.add]
    final: str
```

```python
llm = ChatGroq(model="openai/gpt-oss-120b")
```

```python
def orchestrator(state: State) -> dict:
    
    plan = llm.with_structured_output(Plan).invoke(
        [
            SystemMessage(content= """Create a blog plan with 5-7 sections. "
            "Assign each section a unique sequential ID starting from 1. "
            "Order the sections logically."""),
            
            HumanMessage(content= f"topic: {state['topic']}")
        ]
    )
    
    
    
    return {"plan" : plan}
```

```python
def fanout(state: State):
    return [Send("worker", {"task": task, "topic": state["topic"], "plan" : state["plan"]})
    for task in state["plan"].tasks]
```

```python
def worker(payload: dict) -> dict:
    
    task = payload["task"]
    topic = payload["topic"]
    plan = payload["plan"]
    
    
    blog_title = plan.blog_title
    
    section_md = llm.invoke(
        [
            SystemMessage(content= "Write clean markdown section."),
            HumanMessage(
                content=(
                    f"Blog: {blog_title} \n"
                    f"Topic: {topic} \n"
                    f"Section: {task.title} \n"  
                    f"Brief : {task.brief} \n"
                    "Return only the section content in markdown."
            ))
        ]
    ).content.strip()
    
    return {"sections" : [section_md]}
```

```python
from pathlib import Path

def reducer(state: State) -> dict:
    
    title = state['plan'].blog_title 
    body = "\n\n".join(state['sections']).strip()
    
    final_md = f"# {title}\n\n{body}\n"
    
    filename = title.lower().replace(" ", "_") + ".md"
    Path(filename).write_text(final_md, encoding = "utf-8")
    
    return {"final" : final_md}


# from pathlib import Path

# def reducer(state: State) -> dict:
#     title = state["plan"].blog_title
#     body = "\n\n".join(state["sections"]).strip()

#     final_md = f"# {title}\n\n{body}\n"

#     # Save in the current working directory
#     output_path = Path.cwd() / (
#         title.lower().replace(" ", "_") + ".md"
#     )

#     # Create file and write content
#     output_path.parent.mkdir(parents=True, exist_ok=True)
#     output_path.write_text(final_md, encoding="utf-8")

#     print("File saved at:", output_path.resolve())
#     print("Content length:", len(final_md))
#     print("File size:", output_path.stat().st_size, "bytes")

#     return {"final": final_md}
```

```python
g = StateGraph(State)
g.add_node("orchestrator", orchestrator)
g.add_node("worker", worker)
g.add_node("reducer", reducer)
```

```python
g.add_edge(START, "orchestrator")
g.add_conditional_edges("orchestrator", fanout, ["worker"])
g.add_edge("worker", "reducer")
g.add_edge("reducer", END)
```

```python
app = g.compile()
app
```

```python
out = app.invoke({"topic": "write a blog on self attention"})
out["final"]
```
